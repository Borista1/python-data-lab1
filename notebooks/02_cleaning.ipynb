```python
import sys, os
sys.path.append(os.path.abspath(".."))

import pandas as pd
from IPython.display import display
from src import config
from src.io_utils import read_csv_auto
from src.cleaning import (
    clean_column_names, join_datasets, convert_types,
    replace_hidden_missing, handle_missing_values,
    remove_duplicates, normalize_categories_and_text,
    check_business_rules, detect_outliers
)

# 1. Чтение и очистка имен столбцов
cities_df = clean_column_names(read_csv_auto("../" + config.RAW_CITIES))
pop_df = clean_column_names(read_csv_auto("../" + config.RAW_POPULATION))
print("Исходно строк в городах:", len(cities_df))

# 2. Соединение таблиц
df = join_datasets(cities_df, pop_df)
print("После соединения:", len(df))

# 3. Приведение типов и обработка пропусков
df = convert_types(df)
df = replace_hidden_missing(df)
df = handle_missing_values(df)
print("После обработки пропусков:", len(df))

# 4. Дубликаты, категории, бизнес-правила и выбросы
df = remove_duplicates(df)
print("После удаления дубликатов:", len(df))

df = normalize_categories_and_text(df)
df = check_business_rules(df)
df = detect_outliers(df)
print("Итоговое число строк:", len(df))
display(df.head())

# 5. ПРОВЕРКА СОГЛАСОВАННОСТИ С МОДУЛЕМ src/pipeline.py
df_from_pipeline = pd.read_parquet("../" + config.CLEAN_PARQUET)
is_equal = df.reset_index(drop=True).equals(df_from_pipeline.reset_index(drop=True))
print("Совпадает ли результат блокнота и модуля pipeline.py построчно? ->", is_equal)
```
