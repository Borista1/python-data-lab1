```python
!python --version
!pip install -q pandas polars pyarrow

import pandas as pd
import polars as pl
import matplotlib.pyplot as plt
from google.colab import drive

print("pandas:", pd.__version__, "| polars:", pl.__version__)

# Подключаем Google Диск, где уже лежит твоя папка DATASET
drive.mount('/content/drive')

df = pd.read_csv('/content/drive/MyDrive/DATASET/worldcities.csv', sep=';')
print("Строк и столбцов:", df.shape)
display(df.dtypes.to_frame("Тип"))
display((df.isna().mean() * 100).round(2).to_frame("Пропуски, %"))

# График топ-10 стран по числу городов в базе
df['country'].value_counts().head(10).plot(kind='bar', title='Топ-10 стран по числу городов в датасете')
plt.ylabel('Количество городов')
plt.show()
```
